```python
#Importing the Libraries
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
Dataframe=pd.read_csv('DataSets/Salary_Data.csv')
X=Dataframe.iloc[:,:-1].values
Y=Dataframe.iloc[:,-1].values
Y=np.array(Y).reshape(-1,1)
```

```python
#split the data in train and test sets and apply standardscalar aka zscore normalizzation
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,train_size=0.8,random_state=10)
 # now applying standard scalar 
# scaler=StandardScaler()
# X_train=scaler.fit_transform(X_train)
# X_test=scaler.transform(X_test)
# Y_train=scaler.fit_transform(Y_train)
# Y_test=scaler.transform(Y_test)
print(Y_train)


```

```python
#traning data
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(X_train,Y_train)
Ypred=regressor.predict(X)

Error=Ypred-Y
weight,bias=regressor.coef_,regressor.intercept_

def calculateY(X):
    return (X*weight)+bias


#datavisualization
plt.plot(X_train,regressor.predict(X_train),c='g') #model line trained on training x and y



plt.scatter(X,Y,c='r') #orginal datapoints
plt.xlabel("Years of Experince") #X label
plt.ylabel("Salary in NPR") #ylabel
plt.plot(X,Y,c='b') #dataplotting

```
